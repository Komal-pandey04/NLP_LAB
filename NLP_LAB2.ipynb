```python
from nltk.corpus import wordnet as wn
import pandas as pd

# Selected English words
words = [
    "car", "bank", "bat", "dog", "light",
    "plant", "book", "teacher", "student", "computer"
]

data = []

# Explore synsets for every word
for word in words:
    synsets = wn.synsets(word)

    print("\n" + "=" * 60)
    print("WORD:", word)
    print("=" * 60)

    for synset in synsets:
        definition = synset.definition()
        examples = synset.examples()

        print("Synset:", synset.name())
        print("Definition:", definition)

        if examples:
            print("Example:", examples[0])
        else:
            print("Example: No example available")

        print()

        data.append({
            "Word": word,
            "Synset": synset.name(),
            "Definition": definition,
            "Example": examples[0] if examples else "No example"
        })

# Convert results into DataFrame
q1_df = pd.DataFrame(data)

print("\nFinal DataFrame:")
display(q1_df)
```

```python
#Q2 HYPERNYMS
from nltk.corpus import wordnet as wn
import pandas as pd

words = [
    "dog", "cat", "car", "rose",
    "teacher", "computer", "apple",
    "bird", "student", "airplane"
]

results = []

for word in words:

    # Select first noun synset
    synsets = wn.synsets(word, pos=wn.NOUN)

    if not synsets:
        continue

    synset = synsets[0]

    # Direct hypernyms
    direct_hypernyms = synset.hypernyms()

    direct_names = [h.name() for h in direct_hypernyms]

    results.append({
        "Word": word,
        "Synset": synset.name(),
        "Direct Hypernyms": ", ".join(direct_names)
    })

q2_df = pd.DataFrame(results)

display(q2_df)
```

```python
#Trace Hypernym Hierarchy
def hypernym_chain(synset, levels=3):
    """
    Returns hypernyms up to the specified number of levels.
    """

    chain = []
    current = synset

    for _ in range(levels):
        hypernyms = current.hypernyms()

        if not hypernyms:
            break

        current = hypernyms[0]
        chain.append(current.name())

    return chain


words_for_chain = [
    "dog",
    "cat",
    "car",
    "rose",
    "bird"
]

hierarchy_data = []

for word in words_for_chain:

    synsets = wn.synsets(word, pos=wn.NOUN)

    if synsets:
        synset = synsets[0]

        chain = hypernym_chain(synset, levels=3)

        hierarchy_data.append({
            "Word": word,
            "Synset": synset.name(),
            "Hypernym Chain": " → ".join(chain)
        })

hierarchy_df = pd.DataFrame(hierarchy_data)

display(hierarchy_df)
```

```python
#Q3. WordNet Semantic Similarity
from nltk.corpus import wordnet as wn
import pandas as pd

word_pairs = [
    ("car", "automobile"),
    ("dog", "cat"),
    ("dog", "animal"),
    ("dog", "computer"),
    ("car", "vehicle"),
    ("teacher", "student"),
    ("cat", "tiger"),
    ("computer", "laptop")
]


def get_best_similarity(word1, word2):

    synsets1 = wn.synsets(word1, pos=wn.NOUN)
    synsets2 = wn.synsets(word2, pos=wn.NOUN)

    best_score = 0
    best_pair = (None, None)

    for s1 in synsets1:
        for s2 in synsets2:

            score = s1.path_similarity(s2)

            if score is not None and score > best_score:
                best_score = score
                best_pair = (s1.name(), s2.name())

    return best_pair[0], best_pair[1], best_score


results = []

for word1, word2 in word_pairs:

    syn1, syn2, score = get_best_similarity(word1, word2)

    results.append({
        "Word 1": word1,
        "Word 2": word2,
        "Synset 1": syn1,
        "Synset 2": syn2,
        "Path Similarity": score
    })


q3_df = pd.DataFrame(results)

# Rank from highest to lowest similarity
q3_df = q3_df.sort_values(
    by="Path Similarity",
    ascending=False
).reset_index(drop=True)

q3_df.insert(0, "Rank", range(1, len(q3_df) + 1))

display(q3_df)
```

```python
#Q4. Homonyms and Contextual Meaning
from nltk.corpus import wordnet as wn
import pandas as pd

homonyms = ["bank", "bat", "bark", "light", "match"]

# Display different WordNet meanings
for word in homonyms:

    print("\n" + "=" * 60)
    print("WORD:", word)
    print("=" * 60)

    synsets = wn.synsets(word)

    for synset in synsets[:5]:
        print("Synset:", synset.name())
        print("Definition:", synset.definition())
        print()
```
