```python
from nltk.corpus import brown
from nltk.tokenize import sent_tokenize, word_tokenize
from nltk.corpus import stopwords
```

```python
brown_sentences = brown.sents()
```

```python
sentences = [' '.join(sentence) for sentence in brown_sentences[:20]]

print("Total sentences selected:", len(sentences))
```

```python
text = ' '.join(sentences)
```

```python
sentence_tokens = sent_tokenize(text)

print("\\n========== SENTENCE TOKENS ==========")

for i, sentence in enumerate(sentence_tokens, 1):
    print(f"{i}. {sentence}")
```

```python
word_tokens = word_tokenize(text)

print("\\n========== WORD TOKENS ==========")
print(word_tokens[:100])
```

```python
lower_tokens = [word.lower() for word in word_tokens]

print("\\n========== LOWERCASE TOKENS ==========")
print(lower_tokens[:100])
```

```python
stop_words = set(stopwords.words('english'))
```

```python
cleaned_tokens = [
    word for word in lower_tokens
    if word.isalpha() and word not in stop_words
]

print("\\n========== CLEANED TOKENS ==========")
print(cleaned_tokens[:100])
```

```python
print("\\n\\n========== ORIGINAL VS CLEANED TOKENS ==========")

for i, sentence in enumerate(sentence_tokens[:5], 1):

    original = word_tokenize(sentence)

    cleaned = [
        word.lower()
        for word in original
        if word.isalpha() and word.lower() not in stop_words
    ]

    print("\\n----------------------------------------")
    print("Sentence", i)
    print("Original Sentence:")
    print(sentence)

    print("\\nOriginal Tokens:")
    print(original)

    print("\\nCleaned Tokens:")
    print(cleaned)
```

```python
import nltk


nltk.download('brown')
nltk.download('wordnet')
nltk.download('omw-1.4')


from nltk.corpus import brown
from nltk.stem import PorterStemmer
from nltk.stem import WordNetLemmatizer
import pandas as pd
```

```python
brown_words = brown.words()

alphabetic_words = [
    word.lower()
    for word in brown_words
    if word.isalpha()
]

words = alphabetic_words[:30]

print("First 30 alphabetic words:")
print(words)
```

```python
required_words = [
    'playing',
    'studies',
    'running',
    'better',
    'cars'
]

words.extend(required_words)

print("\\nWords used for analysis:")
print(words)
```

```python
stemmer = PorterStemmer()

stemmed_words = [
    stemmer.stem(word)
    for word in words
]
```

```python
lemmatizer = WordNetLemmatizer()

lemmatized_words = [
    lemmatizer.lemmatize(word)
    for word in words
]
```

```python
df = pd.DataFrame({
    'Original Word': words,
    'Stemmed Word': stemmed_words,
    'Lemmatized Word': lemmatized_words
})

print("\\n========== STEMMING VS LEMMATIZATION ==========")

display(df)
```

```python
print("\\nPOS-aware Lemmatization:")

print("playing ->",
      lemmatizer.lemmatize("playing", pos="v"))

print("studies ->",
      lemmatizer.lemmatize("studies", pos="v"))

print("running ->",
      lemmatizer.lemmatize("running", pos="v"))

print("better ->",
      lemmatizer.lemmatize("better", pos="a"))

print("cars ->",
      lemmatizer.lemmatize("cars", pos="n"))
```

```python
required_words = [
    'playing',
    'studies',
    'running',
    'better',
    'cars'
]

stemmed = [
    stemmer.stem(word)
    for word in required_words
]

lemmatized = [
    lemmatizer.lemmatize('playing', pos='v'),
    lemmatizer.lemmatize('studies', pos='v'),
    lemmatizer.lemmatize('running', pos='v'),
    lemmatizer.lemmatize('better', pos='a'),
    lemmatizer.lemmatize('cars', pos='n')
]

comparison_df = pd.DataFrame({
    'Original Word': required_words,
    'Porter Stem': stemmed,
    'WordNet Lemma': lemmatized
})

display(comparison_df)
```
